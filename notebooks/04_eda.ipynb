```python
df["order_delivered_customer_date"] = pd.to_datetime(
    df["order_delivered_customer_date"]
)

df["order_estimated_delivery_date"] = pd.to_datetime(
    df["order_estimated_delivery_date"]
)

delivery_data = df.dropna(
    subset=[
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
).copy()

delivery_data["delivery_days"] = (
    delivery_data["order_delivered_customer_date"]
    - delivery_data["order_purchase_timestamp"]
).dt.days

delivery_data["delivery_delay_days"] = (
    delivery_data["order_delivered_customer_date"]
    - delivery_data["order_estimated_delivery_date"]
).dt.days
```

```python
delivery_data[
    ["delivery_days", "delivery_delay_days"]
].describe()
```

```python
late_delivery_rate = (
    delivery_data["delivery_delay_days"] > 0
).mean() * 100

late_delivery_rate
```

```python
delivery_data["delivery_delay_days"].value_counts().sort_index().tail(10)
```

```python
reviews = pd.read_csv("../data/processed/reviews.csv")

order_reviews = (
    reviews.groupby("order_id")
    .agg(review_score=("review_score", "mean"))
    .reset_index()
)

delivery_reviews = delivery_data.merge(
    order_reviews,
    on="order_id",
    how="left",
    validate="many_to_one"
)
```

```python
delivery_reviews.groupby(
    delivery_reviews["delivery_delay_days"] > 0
)["review_score"].mean()
```

```python
review_distribution = (
    reviews["review_score"]
    .value_counts()
    .sort_index()
)

review_distribution
```

```python
category_reviews = (
    delivery_reviews
    .groupby("product_category_name_english")["review_score"]
    .mean()
    .sort_values()
)

category_reviews.head(10)
```

```python
category_review_stats = (
    delivery_reviews
    .groupby("product_category_name_english")
    .agg(
        avg_review_score=("review_score", "mean"),
        review_count=("review_score", "count")
    )
    .sort_values("review_count", ascending=False)
)

category_review_stats.head(10)
```

```python
seller_performance = (
    df.groupby("seller_id")
      .agg(
          total_sales=("price", "sum"),
          order_items=("order_item_id", "count"),
          avg_item_price=("price", "mean")
      )
      .sort_values("total_sales", ascending=False)
)

seller_performance.head(10)
```

```python
order_delivery = (
    delivery_data.groupby("order_id")
    .agg(
        delivery_delay_days=("delivery_delay_days", "first")
    )
    .reset_index()
)
```

```python
seller_order_data = (
    df.groupby(["seller_id", "order_id"])
      .agg(
          order_value=("price", "sum")
      )
      .reset_index()
)

seller_order_data = seller_order_data.merge(
    order_reviews,
    on="order_id",
    how="left",
    validate="many_to_one"
)

seller_order_data = seller_order_data.merge(
    order_delivery,
    on="order_id",
    how="left",
    validate="many_to_one"
)

seller_performance = (
    seller_order_data
    .groupby("seller_id")
    .agg(
        total_sales=("order_value", "sum"),
        avg_review_score=("review_score", "mean"),
        late_delivery_rate=("delivery_delay_days",
                            lambda x: (x > 0).mean() * 100),
        order_count=("order_id", "nunique")
    )
    .sort_values("total_sales", ascending=False)
)

seller_performance.head(10)
```
